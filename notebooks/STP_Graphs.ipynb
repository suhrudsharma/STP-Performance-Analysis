```python

# Cell 1: Imports and output setup

from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

GRAPH_DIR = Path("results/graphs")
GRAPH_DIR.mkdir(parents=True, exist_ok=True)

print("STP visualization environment ready.")
print("Graph output directory:", GRAPH_DIR)
```

```python

# Cell 2: Load processed measurements

def load_processed_stp_data(file_path):
    """Load processed STP measurements from CSV."""

    path = Path(file_path)

    if not path.exists():
        raise FileNotFoundError(
            f"Processed dataset not found: {path}"
        )

    df = pd.read_csv(path)

    if "topology" not in df.columns:
        raise ValueError("Dataset must contain a topology column.")

    print("Loaded:", path)
    print("Rows:", len(df))

    return df


print("Graph data loader defined successfully.")
```

```python

# Cell 3: Reusable graph function

def plot_topology_metric(
    df,
    metric,
    title,
    ylabel,
    filename,
    aggregate="mean",
):
    """Plot an available numeric metric by topology."""

    if "topology" not in df.columns or metric not in df.columns:
        raise ValueError(
            f"Required columns missing: topology, {metric}"
        )

    data = df[["topology", metric]].copy()
    data[metric] = pd.to_numeric(data[metric], errors="coerce")
    data = data.dropna(subset=["topology", metric])

    if data.empty:
        raise ValueError(f"No valid measurements for {metric}.")

    grouped = data.groupby("topology")[metric].agg(aggregate)

    ax = grouped.plot(kind="bar", figsize=(9, 5))
    ax.set_title(title)
    ax.set_xlabel("Topology")
    ax.set_ylabel(ylabel)
    ax.tick_params(axis="x", rotation=25)

    plt.tight_layout()
    output_path = GRAPH_DIR / filename
    plt.savefig(output_path, dpi=300, bbox_inches="tight")
    plt.show()

    print("Saved graph:", output_path)


print("Graph plotting function defined successfully.")
```

```python

# Cell 4: Convergence time comparison

def plot_convergence(df):
    plot_topology_metric(
        df=df,
        metric="convergence_time_ms",
        title="STP Convergence Time by Topology",
        ylabel="Convergence Time (ms)",
        filename="convergence_time.png",
    )


print("Convergence plotting function defined successfully.")
```

```python

# Cell 5: Packet loss comparison

def plot_packet_loss(df):
    plot_topology_metric(
        df=df,
        metric="packet_loss_percent",
        title="Packet Loss by Topology",
        ylabel="Packet Loss (%)",
        filename="packet_loss.png",
    )


print("Packet-loss plotting function defined successfully.")
```

```python

# Cell 6: Downtime comparison

def plot_downtime(df):
    plot_topology_metric(
        df=df,
        metric="downtime_ms",
        title="Service Downtime by Topology",
        ylabel="Downtime (ms)",
        filename="downtime.png",
    )


print("Downtime plotting function defined successfully.")
```

```python

# Cell 7: Port state comparisons

def plot_port_states(df):
    required = {"topology", "blocked_ports", "forwarding_ports"}

    if not required.issubset(df.columns):
        raise ValueError(
            "Dataset must contain topology, blocked_ports, "
            "and forwarding_ports."
        )

    data = df.copy()

    for col in ["blocked_ports", "forwarding_ports"]:
        data[col] = pd.to_numeric(data[col], errors="coerce")

    summary = data.groupby("topology")[
        ["blocked_ports", "forwarding_ports"]
    ].mean()

    ax = summary.plot(kind="bar", figsize=(9, 5))
    ax.set_title("Blocked and Forwarding Ports by Topology")
    ax.set_xlabel("Topology")
    ax.set_ylabel("Average Port Count")
    ax.tick_params(axis="x", rotation=25)

    plt.tight_layout()
    path = GRAPH_DIR / "port_states.png"
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()

    print("Saved graph:", path)


print("Port-state plotting function defined successfully.")
```

```python

# Cell 8: BPDU count comparison

def plot_bpdu_count(df):
    plot_topology_metric(
        df=df,
        metric="bpdu_count",
        title="Observed BPDU Count by Topology",
        ylabel="BPDU Count per Defined Capture Interval",
        filename="bpdu_count.png",
    )


print("BPDU plotting function defined successfully.")
```

```python

# Cell 9: Path-cost comparison

def plot_path_cost(df):
    plot_topology_metric(
        df=df,
        metric="path_cost",
        title="Observed Path Cost by Topology",
        ylabel="Path Cost",
        filename="path_cost.png",
    )


print("Path-cost plotting function defined successfully.")
```
